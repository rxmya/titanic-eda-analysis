# Step 11: Data Cleaning

Data cleaning is performed to prepare the dataset for further analysis.

The dataset contains missing values in the `Age`, `Cabin`, and `Embarked` columns. These missing values need to be handled before performing detailed analysis.

The following approaches are used:

- Missing values in `Age` are replaced with the median age.
- Missing values in `Embarked` are replaced with the most frequent value (mode).
- The `Cabin` column is removed because a large proportion of its values are missing.

## 11.1 Handling Missing Age Values

The missing values in the `Age` column are replaced using the median age.

The median is preferred because it is less affected by extreme values or outliers compared with the mean.

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

## 11.2 Handling Missing Embarked Values

The `Embarked` column contains only a small number of missing values.

These missing values are replaced with the mode, which represents the most frequently occurring embarkation port.

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

## 11.3 Removing the Cabin Column

The `Cabin` column contains a very high percentage of missing values.

Since most of the values are unavailable and the column is not essential for the planned analysis, it is removed from the dataset.

```python
df.drop("Cabin", axis=1, inplace=True)
```

## 11.4 Verifying the Data Cleaning

The missing values are checked again to confirm that the required missing-value treatment has been completed.

```python
df.isnull().sum()
```

### Observation

The missing values in the `Age` column were replaced using the median, while the missing values in `Embarked` were replaced using the mode. The `Cabin` column was removed because it contained a large proportion of missing values.

The dataset is now ready for exploratory analysis.

# Step 13: Bivariate Analysis

Bivariate Analysis involves studying the relationship between two variables.

In this step, survival is compared with different passenger characteristics to identify factors that may be associated with survival.

The following relationships are analyzed:

- Gender and survival
- Passenger class and survival
- Age group and survival
- Fare and survival

## 13.1 Survival by Gender

The relationship between gender and survival is analyzed to determine whether survival rates differed between male and female passengers.

### Cross-Tabulation

A cross-tabulation is used to compare the number of passengers who survived and did not survive across different genders.

```python
pd.crosstab(df["Sex"], df["Survived"])
```

### Survival Rate by Gender

The survival rate for each gender is calculated by grouping passengers according to gender and calculating the mean of the `Survived` column.

Since `Survived` contains 0 and 1 values, its mean represents the proportion of passengers who survived.

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean() * 100
survival_by_gender
```

### Visualization

A bar plot is used to compare the survival rates between male and female passengers.

```python
plt.figure(figsize=(6,4))

sns.barplot(
    x="Sex",
    y="Survived",
    data=df
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

### Observation
Female passengers had a significantly higher survival rate than male passengers. Approximately 74% of female passengers survived compared with about 19% of male passengers. This indicates that gender was strongly associated with survival in the Titanic dataset.

## 13.2 Survival by Passenger Class

The relationship between passenger class and survival is analyzed to determine whether passengers in different classes had different survival rates.

```python
pd.crosstab(df["Pclass"], df["Survived"])
```

### Survival Rate by Passenger Class

The survival rate for each passenger class is calculated to provide a fair comparison between the three classes.

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
survival_by_class
```

### Visualization

A bar plot is used to compare the survival rates of first-, second-, and third-class passengers.

```python
plt.figure(figsize=(6,4))

sns.barplot(
    x="Pclass",
    y="Survived",
    data=df
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.show()
```

### Observation

Passenger class had a strong relationship with survival. First-class passengers had the highest survival rate at approximately 63%, followed by second-class passengers at approximately 47%. Third-class passengers had the lowest survival rate at approximately 24%.

## 13.3 Survival by Age Group

The relationship between age and survival is analyzed by creating age groups.

Grouping passengers into age categories makes it easier to compare survival patterns across different stages of age.

### Creating Age Groups

The `Age` column is divided into four categories:

- Child: 0–12 years
- Teenager: 13–17 years
- Adult: 18–59 years
- Senior: 60 years and above

```python
bins = [0, 12, 17, 59, 100]

labels = ["Child", "Teenager", "Adult", "Senior"]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

### Checking the Age Groups

The number of passengers in each age group is checked to verify that the new categories have been created correctly.

```python
df["AgeGroup"].value_counts()
```

### Survival Rate by Age Group

The survival rate is calculated for each age group to compare survival patterns across different age categories.

```python
survival_by_age = df.groupby("AgeGroup", observed=True)["Survived"].mean() * 100
survival_by_age
```

### Visualization

A bar plot is used to compare the survival rates across the different age groups.

```python
plt.figure(figsize=(7,4))

sns.barplot(
    x="AgeGroup",
    y="Survived",
    data=df
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")

plt.ylim(0, 1)

plt.show()
```

### Observation

The survival rate varied across age groups. Children showed a different survival pattern compared with adults and seniors, indicating that age was also associated with survival outcomes.

## 13.4 Fare Distribution by Survival

The relationship between ticket fare and survival is explored using a box plot.

A box plot is useful for comparing the distribution of a numerical variable across different groups.

```python
plt.figure(figsize=(7,4))

sns.boxplot(
    x="Survived",
    y="Fare",
    data=df
)

plt.title("Fare Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Fare")

plt.show()
```

### Observation

Passengers who survived generally had higher ticket fares compared with passengers who did not survive. This suggests that fare, which is related to passenger class, may have been associated with survival.

## 14.1 Selecting Numerical Variables

Correlation is calculated between numerical variables. Therefore, the relevant numerical columns are selected from the dataset.

```python
numeric_df = df.select_dtypes(include=np.number)
numeric_df.head()
```

## 14.2 Correlation Matrix

A correlation matrix is created to calculate the correlation coefficient between every pair of numerical variables.

The matrix helps identify positive and negative relationships between variables.

```python
correlation_matrix = numeric_df.corr()
correlation_matrix
```

## 14.3 Correlation Heatmap

A heatmap is used to visually represent the correlation matrix.

The numerical values are displayed inside the cells, making it easier to identify stronger positive and negative relationships between variables.

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The correlation analysis shows noticeable relationships between passenger class, fare, and survival. Passenger class has a negative relationship with survival because higher class numbers represent lower passenger classes. Fare has a positive relationship with survival, indicating that passengers who paid higher fares generally had better survival outcomes. However, correlation does not imply causation.
